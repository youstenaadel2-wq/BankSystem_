```python
from datetime import datetime

class Customer:
    def __init__(self, customer_id, name, phone, address):
        self.customer_id = customer_id
        self.name = name
        self.phone = phone
        self.address = address

    def display(self):
        return (f"Customer ID : {self.customer_id}\n"
                f"Name : {self.name}\n"
                f"Phone : {self.phone}\n"
                f"Address : {self.address}" )

        def __str__(self):
            return (f"Customer ID : {self.customer_id}\n"
                    f"Name : {self.name}\n"
                    f"Phone : {self.phone}\n"
                    f"Address : {self.address}" )

    def save_customer(self):
        try:
            with open("customers.txt","a") as file:
                file.write(
                    f"{self.customer_id},"
                    f"{self.name},"
                    f"{self.phone},"
                    f"{self.address}\n"
                )
        except Exception as e:
            print("Customer Save Error:",e)
```

```python
class Account:
    def __init__(self,account_number,customer_id,account_type,balance,pin):
        self.account_number = account_number
        self.customer_id = customer_id
        self.account_type = account_type
        self.balance = float(balance)
        self.__pin = pin
        self.transactions = []

    def display(self):
        return(
            f"Account Number : {self.account_number}\n"
            f"Customer ID : {self.customer_id}\n"
            f"Account Type : {self.account_type}\n"
            f"Balance : {self.balance}")

    def __str__(self):
       return (
        f"Account Number : {self.account_number}\n"
        f"Customer ID : {self.customer_id}\n"
        f"Account Type : {self.account_type}\n"
        f"Balance : {self.balance}")

    def check_pin(self,pin):
        if pin==self.__pin:
            return True
        return False

    def change_pin(self,old_pin,new_pin):
        try:
            if old_pin!=self.__pin:
                raise Exception("Wrong PIN")
            if len(new_pin)!=4 or not new_pin.isdigit():
                raise Exception("PIN Must Be 4 Digits")
            self.__pin=new_pin
            print("PIN Changed Successfully")
            return True
        except Exception as e:
            print(e)
            return False

    def deposit(self,amount):
        try:
            amount=float(amount)
            if amount<=0:
                raise Exception("Invalid Amount")
            self.balance+=amount
            transaction=Transaction(self.account_number,"Deposit",amount)
            self.transactions.append(transaction)
            print("Deposit Successful")
            return True
        except Exception as e:
            print(e)
            return False

    def withdraw(self,amount):
        try:
            amount=float(amount)
            if amount<=0:
                raise Exception("Invalid Amount")
            if amount>self.balance:
                raise Exception("Insufficient Balance")
            self.balance-=amount
            transaction=Transaction(self.account_number,"Withdraw",amount)
            self.transactions.append(transaction)
            print("Withdraw Successful")
            return True
        except Exception as e:
            print(e)
            return False

    def transfer(self,receiver,amount):
        try:
            amount=float(amount)
            if self.account_number==receiver.account_number:
                raise Exception("Cannot Transfer To Same Account")
            if amount<=0:
                raise Exception("Invalid Amount")
            if amount>self.balance:
                raise Exception("Insufficient Balance")
            self.balance-=amount
            receiver.balance+=amount
            sender_transaction=Transaction(self.account_number,f"Transfer To {receiver.account_number}",amount)
            self.transactions.append(sender_transaction)
            receiver_transaction=Transaction(receiver.account_number,f"Transfer From {self.account_number}",amount)
            receiver.transactions.append(receiver_transaction)
            print("Transfer Successful")
            return True
        except Exception as error:
            print(error)
            return False

    def add_transaction(self,operation,amount):
        transaction=Transaction(self.account_number,operation,amount)
        self.transactions.append(transaction)

    def show_transactions(self):
        if len(self.transactions)==0:
            print("No Transactions Found")
            return
        print("Transactions")
        for transaction in self.transactions:
            print(transaction.display())
            print("-"*40)

    def get_pin(self):
        return self.__pin
```

```python

class Transaction:
    def __init__(self,account_number,operation,amount):
           self.account_number = account_number
           self.operation = operation
           self.amount = float(amount)
           self.date = datetime.now()
    def display(self):
        return (

            f"Account : {self.account_number}\n"
            f"Operation : {self.operation}\n"
            f"Amount : {self.amount}\n"
            f"Date : {self.date}"
        )
    def __str__(self):
       return (
        f"Account Number : {self.account_number}\n"
        f"Customer ID : {self.customer_id}\n"
        f"Account Type : {self.account_type}\n"
        f"Balance : {self.balance}"
    )

    def save_transaction(self):
        try:
            with open( "transactions.txt","a") as file:
                file.write(
                    f"{self.account_number},"
                    f"{self.operation},"
                    f"{self.amount},"
                    f"{self.date}\n"
                )
        except Exception as e:
            print("Transaction Save Error:",e)
```

```python
from datetime import datetime

class Filemanager:
    def save_customer(self, customer):
        try:
            with open("customers.txt","a") as file:
                file.write(
                    f"{customer.customer_id},"
                    f"{customer.name},"
                    f"{customer.phone},"
                    f"{customer.address}\n"
                )
        except Exception as e:
            print("Customer Save Error :",e)

    def load_customers(self):
        customers={}
        try:
            with open("customers.txt","r") as file:
                for line in file:
                    line=line.strip()
                    if line=="":
                        continue
                    data=line.split(",")
                    customer=Customer(data[0],data[1],data[2],data[3])
                    customers[customer.customer_id]=customer
        except FileNotFoundError:
            pass
        except Exception as e:
            print("Load Customer Error :",e)
        return customers

    def save_accounts(self, accounts):
        try:
            with open("accounts.txt","w") as file:
                for account in accounts.values():
                    file.write(
                        f"{account.account_number},"
                        f"{account.customer_id},"
                        f"{account.account_type},"
                        f"{account.balance},"
                        f"{account.get_pin()}\n"
                    )
        except Exception as e:
            print("Save Account Error :",e)

    def load_accounts(self):
        accounts = {}
        try:
            with open("accounts.txt", "r") as file:
                for line in file:
                    line = line.strip()
                    if not line:
                        continue
                    data = line.split(",")
                    if len(data) == 5:
                        # Account constructor: account_number, customer_id, account_type, balance, pin
                        account = Account(data[0], data[1], data[2], float(data[3]), data[4])
                        accounts[account.account_number] = account
                    else:
                        print(f"Skipping malformed account line: {line}")
        except FileNotFoundError:
            pass
        except Exception as e:
            print("Load Account Error :", e)
        return accounts

    def save_transaction(self, transaction):
        try:
            with open("transactions.txt","a") as file:
                file.write(
                    f"{transaction.account_number},"
                    f"{transaction.operation},"
                    f"{transaction.amount},"
                    f"{transaction.date}\n"
                )
        except Exception as e:
            print("Transaction Save Error :",e)

    def load_transactions(self):
        transactions = []
        try:
            with open("transactions.txt", "r") as file:
                for line in file:
                    line = line.strip()
                    if not line:
                        continue
                    data = line.split(",")
                    # Transaction constructor: account_number, operation, amount
                    # Date is generated by datetime.now() in the constructor, so we don't pass data[3]
                    if len(data) >= 3:
                        transaction = Transaction(data[0], data[1], float(data[2]))
                        # If we want to preserve the date, Transaction.__init__ needs to be adjusted.
                        # For now, new transaction objects will have current datetime.
                        transactions.append(transaction)
                    else:
                        print(f"Skipping malformed transaction line: {line}")
        except FileNotFoundError:
            pass
        except Exception as e:
            print("Load Transaction Error :", e)
        return transactions
```

```python
class BankSystem:
    def __init__(self):
        self.file = Filemanager()
        self.customers = self.file.load_customers()
        self.accounts = self.file.load_accounts()
        self.transactions = self.file.load_transactions()
    def create_customer(self):
        print("\n========== Create Customer ==========")
        customer_id=input("Enter Customer ID : ")
        name=input("Enter Customer Name : ")
        phone=input("Enter Phone Number : ")
        address=input("Enter Address : ")
        customer=Customer(customer_id,name,phone,address)
        self.customers[customer_id]=customer
        self.file.save_customer(customer)
        print("Customer Created Successfully")

    def create_account(self):
        print("\n========== Create Account ==========")
        customer_id=input("Enter Customer ID : ")
        if customer_id not in self.customers:
            print("Customer Not Found")
            return
        account_number=input("Enter Account Number : ")
        account_type=input("Enter Account Type (Saving/Current): ")
        balance=float(input("Enter Initial Balance : "))
        pin=input("Enter PIN : ")
        account=Account(account_number,customer_id,account_type,balance,pin)
        self.accounts[account_number]=account
        self.file.save_accounts(self.accounts)
        print("Account Created Successfully")

    def login(self):
        print("\n========== Login ==========")
        account_number=input("Enter Account Number : ")
        pin=input("Enter PIN : ")
        if account_number in self.accounts:
            account=self.accounts[account_number]
            if account.check_pin(pin):
                print("Login Successful")
                return account
            else:
                print("Wrong PIN")
        else:
            print("Account Not Found")
        return None

    def deposit(self,account):
        amount=float(input("Enter Deposit Amount : "))
        account.deposit(amount)
        transaction=Transaction(account.account_number,"Deposit",amount)
        self.transactions.append(transaction)
        self.file.save_transaction(transaction)
        self.file.save_accounts(self.accounts)
        print("Deposit Completed")

    def withdraw(self,account):
        amount=float(input("Enter Withdraw Amount : "))
        result=account.withdraw(amount)
        if result:
            transaction=Transaction(account.account_number,"Withdraw",amount)
            self.transactions.append(transaction)
            self.file.save_transaction(transaction)
            self.file.save_accounts(self.accounts)
            print("Withdraw Completed")
        else:
            print("Insufficient Balance")

    def search_account(self):
        account_number=input("Enter Account Number : ")
        if account_number in self.accounts:
            account=self.accounts[account_number]
            print(account)
        else:
            print("Account Not Found")

    def delete_account(self):
        account_number=input("Enter Account Number : ")
        if account_number in self.accounts:
            del self.accounts[account_number]
            self.file.save_accounts(self.accounts)
            print("Account Deleted")
        else:
            print("Account Not Found")

    def show_customer_data(self):
        customer_id=input("Enter Customer ID : ")
        if customer_id in self.customers:
            customer=self.customers[customer_id]
            print(customer)
        else:
            print("Customer Not Found")
```
